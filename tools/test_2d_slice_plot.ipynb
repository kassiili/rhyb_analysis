## Test `SlicePlot`

```python
from pathlib import Path
from slice_plot_tools import SlicePlot, SlicePlotConfig
```

```python
sp_cfg = SlicePlotConfig(
    sim_name="mars_example",
    var_name="n_H+sw_ave",
    var_type="scalar",
    var_label="$n$(H$^+_\\mathrm{sw}$) [m$^{-3}$]",
    var_vmin=1e4,
    var_vmax=1e9,
    # smooth_sig=1,
    colormap="inferno",
    log_col_scale=1,

    # --- output ---
    output_path='./test_figures_2d_slice_plot/slice_plot.png',
    # output_path='./test_figures_2d_slice_plot/slice_plot_test.png',
    # output_path='./test_figures_2d_slice_plot/slice_plot_parallel.png',

    # --- figure style ---
    fig_dpi=100,
    fig_resolution=(1920, 1080),

    sim_box_lims=[(-10170e3, 10170e3), (-10170e3, 10170e3), (-10170e3, 10170e3)],
    # axis_lims=[(-8000e3, 8000e3), (-8000e3, 8000e3), (-8000e3, 8000e3)],

    x_unit=3390e3,
    # x_unit=1.0,
    x_unit_label="$R_p$",

    # --- slice points for 3D simulations in format (axis, location in m) ---
    #       - 0 for z-axis, 1 for y and 2 for x.
    slice_points=[(0, 0.0), (1, 0.0), (2, -2.0*3390e3)],
    # slice_points=[(0, -3.0*3390e3), (1, 0.0), (2, 3.0*3390e3)],
    # slice_points=[(2, 2.0*3390e3), (2, 1.0*3390e3), (2, 0.0), (2, -2.0*3390e3)],

    # --- planet disk: (xz plane, xy plane, yz plane) ---
    show_planet=(True, True, False),
    r_planet=3390e3,
    rp_str="$R_p$"
)

sp = SlicePlot(
    Path("../mars_example/sim_data/state00002000.vlsv"),
    sp_cfg
)
sp.render_plot()
```
